```python
import pandas as pd
import numpy as np

model_df = pd.read_csv(
    "../data/processed/neighbourhood_demand.csv"
)

model_df["datetime"] = pd.to_datetime(
    model_df["datetime"]
)

model_df = model_df[
    model_df["datetime"] >= "2013-01-01"
].copy()

model_df.head()
```

```python
model_df["hour"] = model_df["datetime"].dt.hour

model_df["day"] = model_df["datetime"].dt.day

model_df["month"] = model_df["datetime"].dt.month

model_df["weekday"] = model_df["datetime"].dt.weekday

model_df["is_weekend"] = (
    model_df["weekday"].isin([5,6])
).astype(int)
```

```python
model_df.head()
```

```python
def get_season(month):

    if month in [12,1,2]:
        return 0      # Winter

    elif month in [3,4,5]:
        return 1      # Spring

    elif month in [6,7,8]:
        return 2      # Summer

    else:
        return 3      # Autumn

model_df["season"] = (
    model_df["month"]
    .apply(get_season)
)
```

```python
model_df["lag_1"] = (
    model_df["total_demand_kwh"]
    .shift(1)
)
```

```python
model_df["lag_48"] = (
    model_df["total_demand_kwh"]
    .shift(48)
)
```

```python
model_df["lag_96"] = (
    model_df["total_demand_kwh"]
    .shift(96)
)
```

```python
model_df["rolling_mean_48"] = (
    model_df["total_demand_kwh"]
    .rolling(48)
    .mean()
)
```

```python
model_df["rolling_std_48"] = (
    model_df["total_demand_kwh"]
    .rolling(48)
    .std()
)
```

```python
model_df.isnull().sum()
```

```python
model_df = model_df.dropna()
```
