```python
import pandas as pd

df = pd.read_csv(r"C:\Users\tejas\OneDrive\Desktop\DRG_project\data\raw\consumption_d.csv")
```

```python
df["GMT"] = pd.to_datetime(df["GMT"])

df.head()
```

```python
print(df.dtypes.head())
```

```python
print(df["GMT"].min())
print(df["GMT"].max())
```

```python
duplicates = df.duplicated().sum()

print("Duplicates:", duplicates)
```

```python
df = df.drop_duplicates()
```

```python
missing = df.isnull().sum().sum()

print("Total Missing Values:", missing)
```

```python
df_long = df.melt(
    id_vars=["GMT"],
    var_name="household_id",
    value_name="consumption_kwh"
)

df_long.head()
```

```python
df_long.rename(
    columns={"GMT":"datetime"},
    inplace=True
)
```

```python
df_long.shape
```

```python
df_long = df_long.dropna()
```
